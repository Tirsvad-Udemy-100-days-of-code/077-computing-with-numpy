## Lists versus arrays

`+` on two Python lists joins them, and `*` on two lists is an error. Arrays do the maths.

```python
from computing_with_numpy.domain.linear_algebra import (
    VECTOR_1,
    VECTOR_2,
    add_lists,
    multiply_lists,
)

print(add_lists(VECTOR_1, VECTOR_2))
try:
    multiply_lists(VECTOR_1, VECTOR_2)
except TypeError as error:
    print(f'TypeError: {error}')
```

## Matrix multiplication challenge

`a1 @ b1` multiplies a (4, 2) matrix by a (2, 3) matrix and gives a (4, 3) matrix. `np.matmul` gives the same result.

```python
import numpy as np

from computing_with_numpy.domain.linear_algebra import (
    create_matrix_a1,
    create_matrix_b1,
    matrix_product,
)

a1 = create_matrix_a1()
b1 = create_matrix_b1()
c = matrix_product(a1, b1)
print(c)
print(c.shape)
print(np.array_equal(c, np.matmul(a1, b1)))
```

## Convert the image to grayscale

Divide the sRGB values by 255, then the matrix product with the grey values `[0.2126, 0.7152, 0.0722]` gives one channel from 0 to 1. `cmap='gray'` draws it in grey.

```python
import matplotlib.pyplot as plt

from computing_with_numpy.domain.images import load_face_image, to_grayscale

img_gray = to_grayscale(load_face_image())
print(img_gray.shape)
plt.imshow(img_gray, cmap='gray')
plt.show()
```

## Flip the grayscale image

`np.flip` turns the grayscale image upside down.

```python
import matplotlib.pyplot as plt

from computing_with_numpy.domain.images import (
    flip_image,
    load_face_image,
    to_grayscale,
)

img_gray = to_grayscale(load_face_image())
flipped = flip_image(img_gray)
print(flipped.shape)
plt.imshow(flipped, cmap='gray')
plt.show()
```
