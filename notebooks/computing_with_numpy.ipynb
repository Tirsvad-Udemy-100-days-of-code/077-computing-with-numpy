## 1 dimension: a vector

An ndarray is homogeneous: every element has the same type. `shape` and `ndim` describe it; an element is read by index.

## 2 dimensions: a matrix

A 2-dimensional array has two axes: 2 rows and 4 columns. A value needs one index per axis; `:` takes a whole axis.

```python
from computing_with_numpy.domain.ndarray_basics import (
    create_matrix,
    describe_array,
    matrix_element,
    matrix_row,
)

array_2d = create_matrix()
print(array_2d)
print(describe_array(array_2d))
print(matrix_element(array_2d, 1, 2))
print(matrix_row(array_2d, 0))
```
